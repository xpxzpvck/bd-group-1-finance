# 00 — Setup: Catalog & Schemas

This notebook creates the Unity Catalog schemas for the **Finance** team's TPC-H Lakehouse migration (Group Assignment 1).

It creates a dedicated Unity Catalog `group1_finance` with three schemas:

- **bronze** — raw copy of `samples.tpch` (as-is, metadata only)
- **silver** — 3NF with enforced data quality
- **gold** — aggregated tables answering Finance business questions

```python
# Configuration
# Catalog and schema naming convention:
#   group1_finance.<layer>.<table>

CATALOG = "group1_finance"
SCHEMA_BRONZE = "bronze"
SCHEMA_SILVER = "silver"
SCHEMA_GOLD   = "gold"

# Source data
SOURCE_CATALOG = "samples"
SOURCE_SCHEMA  = "tpch"

# TPC-H tables to migrate
TPCH_TABLES = [
    "customer",
    "lineitem",
    "nation",
    "orders",
    "part",
    "partsupp",
    "region",
    "supplier",
]

print(f"Target catalog : {CATALOG}")
print(f"Bronze schema  : {CATALOG}.{SCHEMA_BRONZE}")
print(f"Silver schema  : {CATALOG}.{SCHEMA_SILVER}")
print(f"Gold schema    : {CATALOG}.{SCHEMA_GOLD}")
print(f"Source         : {SOURCE_CATALOG}.{SOURCE_SCHEMA}")
print(f"Tables ({len(TPCH_TABLES)}): {', '.join(TPCH_TABLES)}")
```

```python
# Drop catalog (cascades to all schemas/tables)
spark.sql(f"DROP CATALOG IF EXISTS {CATALOG} CASCADE")
print(f"Dropped  catalog {CATALOG} (if it existed)")
print()

# Create catalog
spark.sql(f"CREATE CATALOG {CATALOG}")
spark.sql(f"COMMENT ON CATALOG {CATALOG} IS 'Group 1 Finance - TPC-H Lakehouse migration'")
print(f"Created  catalog {CATALOG}")
print()

# Create schemas
for schema in [SCHEMA_BRONZE, SCHEMA_SILVER, SCHEMA_GOLD]:
    spark.sql(f"CREATE SCHEMA {CATALOG}.{schema}")
    spark.sql(f"COMMENT ON SCHEMA {CATALOG}.{schema} IS 'Group 1 Finance - {schema.capitalize()} layer for TPC-H Lakehouse migration'")
    print(f"Created  {CATALOG}.{schema}")
```

```python
# Verify
print(f"Schemas in {CATALOG} catalog:")
display(spark.sql(f"SHOW SCHEMAS IN {CATALOG}"))
```
